```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

```

```python
DATA = Path("../data")
FIG = Path("../figures")
FIG.mkdir(parents=True, exist_ok=True)

```

```python
# ======================
# 1. ЗАГРУЗКА ВСЕХ SAMPLE CSV
# ======================
tox = pd.read_csv(DATA / "toxic_comments_sample.csv")
sales = pd.read_csv(DATA / "store_sales.csv", parse_dates=["date"])
fetal = pd.read_csv(DATA / "fetal_sample.csv")

```

```python
print("toxic:", tox.shape)
print("sales:", sales.shape)
print("fetal:", fetal.shape)

```

```python
# Баланс toxic
tox_counts = tox["toxic"].value_counts().sort_index()

plt.figure(figsize=(6, 4))
sns.barplot(
    x=["clean", "toxic"],
    y=tox_counts.values,
    hue=["clean", "toxic"],
    palette="viridis",
    legend=False,
)
plt.title("Toxic comments class counts", fontsize=14, pad=12)
plt.ylabel("Count")
plt.grid(axis="y", linestyle="--", alpha=0.7)
for i, c in enumerate(tox_counts.values):
    plt.text(i, c + 1, str(c), ha="center")
plt.tight_layout()
plt.show()

```

```python
# Классы fetal CTG
fetal_counts = fetal["target"].value_counts().sort_values(ascending=True)

plt.figure(figsize=(8, 4))
sns.barplot(
    x=fetal_counts.index,
    y=fetal_counts.values,
    hue=fetal_counts.index,
    palette="viridis",
    legend=False,
)
plt.title("Fetal target class counts", fontsize=14, pad=12)
plt.ylabel("Count")
plt.xticks(rotation=15)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

```

```python
# Средние продажи по магазинам
mean_sales = sales.groupby("store_id")["sales"].mean()

plt.figure(figsize=(6, 4))
sns.barplot(
    x=mean_sales.index.astype(str),
    y=mean_sales.values,
    hue=mean_sales.index.astype(str),
    palette="viridis",
    legend=False,
)
plt.title("Mean store sales", fontsize=14, pad=12)
plt.xlabel("store_id")
plt.ylabel("Mean sales")
plt.tight_layout()
plt.show()

```

```python
# ======================
# 2. СВОДНАЯ EDA-ПАНЕЛЬ
# ======================
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))

# Toxic
tox["toxic"].value_counts().sort_index().plot(kind="bar", ax=axes[0], color="#4c72b0", edgecolor="black")
axes[0].set_title("Toxic labels")
axes[0].set_xlabel("toxic")
axes[0].tick_params(axis="x", rotation=0)

# Fetal
fetal["target"].value_counts().plot(kind="bar", ax=axes[1], color="#55a868", edgecolor="black")
axes[1].set_title("Fetal classes")
axes[1].tick_params(axis="x", rotation=20)

# Sales
sales.groupby("store_id")["sales"].mean().plot(kind="bar", ax=axes[2], color="#c44e52", edgecolor="black")
axes[2].set_title("Mean sales / store")
axes[2].set_xlabel("store_id")
axes[2].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.savefig(FIG / "competition_eda_panel.png", dpi=120, bbox_inches="tight")
plt.show()

```

```python
# Длина toxic-комментариев
tox = tox.copy()
tox["n_tokens"] = tox["text"].astype(str).str.split().str.len()

plt.figure(figsize=(8, 4))
sns.boxplot(data=tox, x="toxic", y="n_tokens", palette="viridis")
plt.title("Token count by toxicity", fontsize=14, pad=12)
plt.tight_layout()
plt.show()

```

```python
# Временной ряд продаж (store 1)
s1 = sales[sales["store_id"] == 1]

plt.figure(figsize=(11, 3.5))
plt.plot(s1["date"], s1["sales"], color="#4c72b0")
plt.title("Store 1 daily sales", fontsize=14, pad=10)
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python
# Корреляции fetal (числовые)
num = fetal.select_dtypes(include=[np.number])
plt.figure(figsize=(9, 7))
sns.heatmap(num.corr(), cmap="coolwarm", center=0)
plt.title("Fetal numeric correlations", fontsize=13, pad=10)
plt.tight_layout()
plt.show()

```
