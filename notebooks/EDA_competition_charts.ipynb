# Competition EDA charts


Shared EDA charts. PNGs in `figures/`.


```python
import pandas as pd, matplotlib.pyplot as plt
from pathlib import Path
DATA, FIG = Path("../data"), Path("../figures")
tox = pd.read_csv(DATA / "toxic_comments_sample.csv")
sales = pd.read_csv(DATA / "store_sales.csv", parse_dates=["date"])
fetal = pd.read_csv(DATA / "fetal_sample.csv")
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
tox["toxic"].value_counts().plot(kind="bar", ax=axes[0])
fetal["target"].value_counts().plot(kind="bar", ax=axes[1])
sales.groupby("store_id")["sales"].mean().plot(kind="bar", ax=axes[2])
plt.tight_layout(); plt.savefig(FIG / "competition_eda_panel.png", dpi=120); plt.show()
```
